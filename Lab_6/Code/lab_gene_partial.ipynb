## Background

Genes are the basic unit in the DNA and encode blueprints for proteins.  When proteins are synthesized from a gene, the gene is said to "express".  Micro-arrays are devices that measure the expression levels of large numbers of genes in parallel.  By finding correlations between expression levels and phenotypes, scientists can identify possible genetic markers for biological characteristics.

The data in this lab comes from:

https://archive.ics.uci.edu/ml/datasets/Mice+Protein+Expression

In this data, mice were characterized by three properties:
* Whether they had down's syndrome (trisomy) or not
* Whether they were stimulated to learn or not
* Whether they had a drug memantine or a saline control solution.

With these three choices, there are 8 possible classes for each mouse.  For each mouse, the expression levels were measured across 77 genes.  We will see if the characteristics can be predicted from the gene expression levels.  This classification could reveal which genes are potentially involved in Down's syndrome and if drugs and learning have any noticeable effects.


## Load the Data

We begin by loading the standard modules.

```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn import linear_model, preprocessing
```

## Binary Classification for Down's Syndrome

We will first predict the binary class label in `df1['Genotype']` which indicates if the mouse has Down's syndrome or not.
```python
# TODO 4
#   xnames = ...
#   X = ...

xnames = df1.columns[:-4]
X = df1[xnames].values
```

Split the data into training and test with 30% allocated for test.  You can use the train

```python
# TODO 8
#   yhat = ...

yhat = logreg.predict(Xts1)
print(np.mean(yhat == yts))
```

## Cross Validation

To obtain a slightly more accurate result, now perform 10-fold cross validation and measure the average precision, recall and f1-score.  Note, that in performing the cross-validation, you will want to randomly permute the test and training sets using the `shuffle` option.  In this data set, all the samples from each class are bunched together, so shuffling is essential.  Print the mean precision, recall and f1-score and error rate across all the folds.

```python
from sklearn.model_selection import KFold
from sklearn.metrics import precision_recall_fscore_support
nfold = 10
kf = KFold(n_splits=nfold,shuffle=True)

# TODO 11

precision = []
recall = []
f1 = []
err = []

for train, test in kf.split(X):
    Xtr_k, Xts_k = X[train], X[test]
    ytr_k, yts_k = y[1][train], y[1][test]

    scaler = StandardScaler()
    scaler.fit(Xtr_k)
    Xtr1_k = scaler.transform(Xtr_k)
    Xts1_k = scaler.transform(Xts_k)

    logreg_k = linear_model.LogisticRegression(C=1e5, solver='liblinear')
    logreg_k.fit(Xtr1_k, ytr_k)

    yhat_k = logreg_k.predict(Xts1_k)
    p, r, f, s = precision_recall_fscore_support(yts_k, yhat_k, average='binary')
    err_k = np.mean(yhat_k != yts_k)

    precision.append(p)
    recall.append(r)
    f1.append(f)
    err.append(err_k)

print("Precision: %.3f +- %.3f" % (np.mean(precision), np.std(precision)))
print("Recall: %.3f +- %.3f" % (np.mean(recall), np.std(recall)))
print("F1: %.3f +- %.3f" % (np.mean(f1), np.std(f1)))
print("Error: %.3f +- %.3f" % (np.mean(err), np.std(err)))
```

## Multi-Class Classification

Now use the response variable in `df1['class']`.  This has 8 possible classes.  Use the `np.unique` funtion as before to convert this to a vector `y` with values 0 to 7.

```python
# TODO 12
#   y = ...
y = np.unique(df1['class'].values, return_inverse=True)
```

Fit a multi-class logistic model by creating a `LogisticRegression` object, `logreg` and then calling the `logreg.fit` method.

Now perform 10-fold cross validation, and measure the confusion matrix `C` on the test data in each fold. You can use the `confustion_matrix` method in the `sklearn` package.  Add the confusion matrix counts across all folds and then normalize the rows of the confusion matrix so that they sum to one.  Thus, each element `C[i,j]` will represent the fraction of samples where `yhat==j` given `ytrue==i`.  Print the confusion matrix.  You can use the command

    print(np.array_str(C, precision=4, suppress_small=True))
    
to create a nicely formatted print.  Also print the overall mean and SE of the test accuracy across the folds.

```python
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import KFold

# TODO 13

C_total = np.zeros((8, 8))
acc_list = []

for train, test in kf.split(X):
    Xtr_k, Xts_k = X[train], X[test]
    ytr_k, yts_k = y[1][train], y[1][test]

    scaler_k = StandardScaler()
    Xtr_k_scaled = scaler_k.fit_transform(Xtr_k)
    Xts_k_scaled = scaler_k.transform(Xts_k)



    logreg_multi = linear_model.LogisticRegression(C=1e5, solver='liblinear')
    logreg_multi.fit(Xtr_k_scaled, ytr_k)


    yhat_k = logreg_multi.predict(Xts_k_scaled)
    C_total += confusion_matrix(yts_k, yhat_k, labels=range(8))
    acc_list.append(np.mean(yhat_k == yts_k))

C_normalized = C_total / C_total.sum(axis=1, keepdims=True)

print(np.array_str(C_normalized, precision=4, suppress_small=True))

print(f"Mean Test Accuracy: {np.mean(acc_list):.4f}")
print(f"SE of Test Accuracy: {np.std(acc_list)/np.sqrt(nfold):.4f}")
```

Re-run the logistic regression on the entire training data and get the weight coefficients.  This should be a 8 x 77 matrix.  Create a stem plot of the first row of this matrix to see the coefficients on each of the genes.

```python
# TODO 14

scaler_full = StandardScaler()
X_scaled = scaler_full.fit_transform(X)

logreg_multi_full = linear_model.LogisticRegression(C=1e5, solver='liblinear')
logreg_multi_full.fit(X_scaled, y[1])

W_multi = logreg_multi_full.coef_


plt.figure(figsize=(10, 5))
plt.stem(W_multi[0, :])
plt.title("Multi-Class Logistic Regression Weights")
plt.xlabel("Gene Index")
plt.ylabel("Weight")
plt.show()
```

## L1-Regularization

This section is bonus.

In most genetic problems, only a limited number of the tested genes are likely influence any particular attribute.  Hence, we would expect that the weight coefficients in the logistic regression model should be sparse.  That is, they should be zero on any gene that plays no role in the particular attribute of interest.  Genetic analysis commonly imposes sparsity by adding an l1-penalty term.  Read the `sklearn` [documentation](http://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LogisticRegression.html) on the `LogisticRegression` class to see how to set the l1-penalty and the inverse regularization strength, `C`.

Using the model selection strategies from the [housing demo](../unit05_lasso/demo2_housing.ipynb), use K-fold cross validation to select an appropriate inverse regularization strength.  
* Use 10-fold cross validation
* You should select around 20 values of `C`.  It is up to you find a good range.
* Make appropriate plots and print out to display your results
* How does the accuracy compare to the accuracy achieved without regularization.

```python
# TODO 15

C_values = np.logspace(-3, 3, 20)
mean_accs = []

for C_val in C_values:
    acc_fold = []
    for train, test in kf.split(X):
        Xtr_k, Xts_k = X[train], X[test]
        ytr_k, yts_k = y[1][train], y[1][test]

        scaler_k = StandardScaler()
        Xtr_k_scaled = scaler_k.fit_transform(Xtr_k)
        Xts_k_scaled = scaler_k.transform(Xts_k)

        logreg_l1 = linear_model.LogisticRegression(penalty='l1', C=C_val, solver='liblinear')
        logreg_l1.fit(Xtr_k_scaled, ytr_k)

        yhat_k = logreg_l1.predict(Xts_k_scaled)
        acc_fold.append(np.mean(yhat_k == yts_k))

    mean_accs.append(np.mean(acc_fold))

plt.figure(figsize=(8, 5))
plt.semilogx(C_values, mean_accs, marker='o')
plt.xlabel("Inverse Regularization Strength")
plt.ylabel("Mean CV Accuracy")
plt.title("L1-Regularization")
plt.grid(True)
plt.show()

best_idx = np.argmax(mean_accs)
print(f"C value: {C_values[best_idx]:.4f}")
print(f"Accuracy with L1 regularization: {mean_accs[best_idx]:.4f}")
```
